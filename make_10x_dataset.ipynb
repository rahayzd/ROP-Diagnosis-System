```python
import os
import shutil


source_dir = "/content/drive/MyDrive/ROP-Data/remain"
dest_dir = "/content/drive/MyDrive/ROP-Data/10_images_stack"

os.makedirs(dest_dir, exist_ok=True)

image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff']

copied = 0

for filename in os.listdir(source_dir):
    if any(filename.lower().endswith(ext) for ext in image_extensions):
        src_path = os.path.join(source_dir, filename)
        dst_path = os.path.join(dest_dir, filename)
        shutil.copy2(src_path, dst_path)
        copied += 1

print(f"✅ تعداد {copied} تصویر با موفقیت کپی شد به: {dest_dir}")
```

```python
import os
import shutil

input_folder = r'/content/drive/MyDrive/ROP-Data/10_images_stack'
output_base = r'/content/drive/MyDrive/ROP-Data/10_split_dataset'

positive_dir = os.path.join(output_base, 'Positive')
negative_dir = os.path.join(output_base, 'Negative')

os.makedirs(positive_dir, exist_ok=True)
os.makedirs(negative_dir, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')):
        parts = filename.split('_')
        dg_part = next((p for p in parts if p.startswith('DG')), None)

        if dg_part is not None:
            dg_value = int(dg_part[2:])  
            src_path = os.path.join(input_folder, filename)

            if dg_value == 0:
                dst_path = os.path.join(negative_dir, filename)
            else:
                dst_path = os.path.join(positive_dir, filename)

            shutil.copyfile(src_path, dst_path)

print("تقسیم‌بندی عکس‌ها با موفقیت انجام شد.")
```
